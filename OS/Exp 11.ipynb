```python
blocks = [100,500,200,300,600]
process =[212,417,112,426]
for p in process:
    allocated = False
    for i in range(len(blocks)):
        if blocks[i]>=p:
            print(f"process{p}->Blocks{i}allocated")
            blocks[i]-=p
            allocated = True
            break
    if not allocated:
            print(f"process{p}->Not Allocated")
```
