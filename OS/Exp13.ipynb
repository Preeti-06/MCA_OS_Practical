```python
# Disk Scheduling Algorithms: FCFS and SSTF

requests = [55, 18, 98, 155, 38]
head = 50

print("Initial Head Position:", head)
print("Requests:", requests)

# ---------------- FCFS ----------------
print("\nFCFS Scheduling:")
fcfs_head = head
fcfs_order = []

for r in requests:
    fcfs_order.append(r)
    fcfs_head = r

print("Service Order:", fcfs_order)

# ---------------- SSTF ----------------
print("\nSSTF Scheduling:")
sstf_head = head
sstf_order = []
pending = requests.copy()

while pending:
    next_req = min(pending, key=lambda x: abs(x - sstf_head))
    sstf_order.append(next_req)
    sstf_head = next_req
    pending.remove(next_req)

print("Service Order:", sstf_order)
```
