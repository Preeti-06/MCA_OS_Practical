```python
# Simulating Containers as a Scheduler (FCFS)

containers = [
    {"name": "Container1", "cpu_time": 4},
    {"name": "Container2", "cpu_time": 2},
    {"name": "Container3", "cpu_time": 6}
]

current_time = 0

print("Container Scheduling Simulation:\n")

for container in containers:
    print("Running", container["name"])
    print("CPU Time:", container["cpu_time"])
    print("Start Time:", current_time)

    current_time += container["cpu_time"]

    print("End Time:", current_time)
    print("-" * 30)
```
