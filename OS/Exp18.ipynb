```python
total_cpu = 16     # total CPU cores
total_memory = 64  # total memory in GB

containers = [
    {"name": "web", "cpu": 4, "memory": 16},
    {"name": "db", "cpu": 6, "memory": 32},
    {"name": "cache", "cpu": 2, "memory": 8},
    {"name": "worker", "cpu": 4, "memory": 16},
]

available_cpu = total_cpu
available_memory = total_memory

print("Docker Container Scheduling Simulation\n")

for container in containers:
    print(f"Trying to schedule {container['name']} (CPU: {container['cpu']}, Memory: {container['memory']} GB)")
    
    if container['cpu'] <= available_cpu and container['memory'] <= available_memory:
        available_cpu -= container['cpu']
        available_memory -= container['memory']
        print(f"Scheduled {container['name']}")
    else:
        print(f"Cannot schedule {container['name']} due to insufficient resources")
    
    print(f"Available CPU: {available_cpu}, Available Memory: {available_memory} GB\n")

print("Scheduling simulation complete!")
```
