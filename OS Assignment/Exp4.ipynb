```python
#FCFS and SSTf Disk Scheduling
def fcfs_disk(requests, head):
    total = 0
    for r in requests:
        total += abs(head - r)
        head = r
    print("FCFS Total Head Movement:", total)

def sstf_disk(requests, head):
    total = 0
    req = requests.copy()
    while req:
        closest = min(req, key=lambda x: abs(x - head))
        total += abs(head - closest)
        head = closest
        req.remove(closest)
    print("SSTF Total Head Movement:", total)

requests = [82, 170, 43, 140, 24, 16, 190]
head = 50

fcfs_disk(requests, head)
sstf_disk(requests, head)
```
