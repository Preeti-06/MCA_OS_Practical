```python
#Implement CPU scheduling Algorithms(FCFS & SJF)
def fcfs(processes):
    total_wait = 0
    wait = 0
    for p in processes:
        total_wait += wait
        wait += p
    print("FCFS Average Waiting Time:", total_wait / len(processes))

def sjf(processes):
    processes.sort()
    total_wait = 0
    wait = 0
    for p in processes:
        total_wait += wait
        wait += p
    print("SJF Average Waiting Time:", total_wait / len(processes))

processes = [24, 3, 3]
fcfs(processes.copy())
sjf(processes.copy())
```
