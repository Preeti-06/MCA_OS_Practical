```python
def bankers(alloc, maxm, avail):
    n = len(alloc)
    m = len(avail)
    
    need = [[maxm[i][j] - alloc[i][j] for j in range(m)] for i in range(n)]
    finish = [False] * n
    safe_seq = []
    
    for _ in range(n):
        for i in range(n):
            if not finish[i] and all(need[i][j] <= avail[j] for j in range(m)):
                for j in range(m):
                    avail[j] += alloc[i][j]
                safe_seq.append(i)
                finish[i] = True
    
    if len(safe_seq) == n:
        print("System is in SAFE state")
        print("Safe Sequence:", safe_seq)
    else:
        print("System is in UNSAFE state")

alloc = [[0,1,0],[2,0,0],[3,0,2],[2,1,1],[0,0,2]]
maxm = [[7,5,3],[3,2,2],[9,0,2],[2,2,2],[4,3,3]]
avail = [3,3,2]

bankers(alloc, maxm, avail)
```
